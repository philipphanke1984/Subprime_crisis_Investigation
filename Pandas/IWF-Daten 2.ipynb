```python
##keine Duplikate, bereit für melt (Umtausch von Zeilen und Spalten)

jahre = [str(i) for i in range(2001, 2026)]

df_sql = df3.melt(
    id_vars=['country', 'Indicator', 'PRIMARY_DOMESTIC_CURRENCY'],
    value_vars=jahre,
    var_name='year',
    value_name='value'
)

```

```python
df_sql
```

```python
df_sql.dtypes
```

```python
##nochmal die Types checken
##year als INT speichern

df_sql['year'] = df_sql['year'].astype(int)
```

```python
import os
print(os.getcwd())
```

```python
df_sql.to_csv(
    'IWF_MySQL.csv',
    index=False,
    encoding='utf-8',
    na_rep=''
)
```
