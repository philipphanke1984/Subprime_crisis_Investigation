```python
import pandas as pd
import os
```

```python
df = pd.read_csv(r'C:\Users\phili\OneDrive\Desktop\Abschlussarbeit Weiterbildung\datasets\WDI - wichtige Daten\WDICSV.csv', 
                low_memory = False)
```

```python
##Anzahl der Spalten eingrenzen

jahre = [str(i) for i in range(2001, 2016)]

spalten = ['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code'] + jahre
```

```python
df2 = df[spalten].copy()
```

```python
df2.shape
```

```python
df2.loc[(df2['Indicator Name'].str.contains('employment')), :]
```

```python
##es werden jetzt zwei Indikatoren ausgewählt

indikatoren = [
    'FP.CPI.TOTL.ZG',
    'SL.UEM.TOTL.ZS'
]

df3 = df2[
    df2['Indicator Code'].isin(indikatoren)
].copy()
```

```python
df3 = df3.reset_index(drop=True)
```

```python
df3
```

```python
##melt

df4 = pd.melt(
    df3,
    id_vars=['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code'],
    var_name='year',
    value_name='value'
)

df4['year'] = df4['year'].astype(int)

df4.head()
```

```python
##Länderaggregate entfernen

laender = pd.read_csv(r'C:\Users\phili\OneDrive\Desktop\Abschlussarbeit Weiterbildung\datasets\WDI - wichtige Daten\WDICountry.csv')

gueltige_codes = laender.loc[
    laender['Region'].notna(),
    'Country Code'
]

df4 = df4[
    df4['Country Code'].isin(gueltige_codes)
].copy()
```

```python
df4
```

```python
##Länder angleichen

namen = {
    'Czechia': 'Czech Republic',
    'Estonia': 'Estonia, Republic of',
    'Korea, Rep.': 'Korea, Republic of',
    'Latvia': 'Latvia, Republic of',
    'Lithuania': 'Lithuania, Republic of',
    'Netherlands': 'Netherlands, The',
    'Poland': 'Poland, Republic of',
    'Slovenia': 'Slovenia, Republic of',
    'Turkiye': 'Türkiye, Republic of'
}

df4['Country Name'] = df4['Country Name'].replace(namen)
```
